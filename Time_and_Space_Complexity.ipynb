                                                   Time and Space Complexity 

Time and space Complexity measures an algorithm's efficiency by analyzing how its resource usage scales with the size of input. 

The Time and space Complexity uses Big O Notation to describe the worst case growth rates helping developers choose efficient algorithms, 
especially for large datasets.  

Time Complexity counts operations  while space complexity counts memory, both focusing on the input size rather than specific machine speed or memory. 

The execution time grows as theinput size increases.  The example of notation is as follows: 
O(1) : Time stays the same like the example of accessing an array element 

O (log n) : Time grows slowly  
O (n) :  Time grows proportionally. 
O (n ^2) : Time grows by the square ( nested loops like bubble sort)

Time Complexity is basically that how number of operations is proportional to the input size. 

Time Complexity measures the bumber of operations an algorithm performs. The goal is to express howw the execution time grows. 

1. Identify the input size(n) : 
Determine what constitutes the input size. Foran array or list, n is ususlly the number of elements. For a loop, it might be the number of iterations. 

2. ount Fundamental Operaations : Estimate how many times basic operations are executed. 

3. Analyze Growth Rate : 

a.  Constsnt Time(O(n)) :  The number of operations remains the same egardless of the input size.  
b. Linear Time : The number of operationsgrows proportionally, with the inout size, (Eg. iterating through an array once) 
c. Quadratic Time (O(n^2)) : The number of operations grows with the aquare of the input size (nested loops iterating through same  array.)   
d. Logarithmic Time (O (logn)) : The inpt size is halved in each step (eg. binary search) 
e. Exponential Time (O (2^n)) : The number of operations doubles with each addition to the input size


Space Complexity : Space Complexity measures the amount of memeory an algorithm uses relative to its input size.  

1. Identify Memory Usage : Look for variables, data structures or recursive cell stacks that consume memory. 

2. Analyze Growth Rate : Similar to time comlexity, express how the memory usage grows with input size(n) 

a. Constant Space (O(n)) : The amount of memeory used is fixed, regardless of input size. 
b. Linear Space :  The memory usage grows proportionally with input size. 
c. Quadratic Space : Memory usage grows with the square of input size
