Implement a Python package structure for a project named ecommerce, containing modules for product
management and order processing ? 

```python
from .product import Product
from .inventory import Inventory

class Product : 
    def __init__(self, product_id, name, price): 
        self.product_id = product_id 
        self.name = name 
        self.price = price 

    def __str__(self): 
        return f"{self.name} (Rs {self.price})" 
    
class Inventory : 
    def __init__(self): 
        self.products = {} 

    def add_product(self, product, quantity): 
        self.products[product.product_id] = { 
            "product" : product, 
            "quantity" : quantity
        }

    def is_available (self, product_id, quantity): 
        return (
            product_id in self.products and 
            self.products[product_id]["quantity"] >= quantity
        )
    

class Order : 
    def __init__(self, order_id): 
        self.order_id =order_id 
        self.items = [] 

    def add_item(self, product, quantity): 
        self.items.append((product,quantity)) 

    def get_total(self): 
        return sum(product.price *quantity for product, quantity in self.items) 
    
    def process_paymment(amount): 
        print(f"Payment of Rs{amount} processed successfully") 

from product import Product, Inventory
from order import Order 
from order.payment import process_payment 

p1 = Product(101, "Laptop", 50000)
inventory = Inventory()
inventory.add_product(p1, 10)

order = Order(1)
order.add_item(p1, 1)

total = order.get_total()
process_payment(total)
```

```python
#  the code for the aove example is as : 

def reverse_string(text): 

    if not isinstance(text, str): 
        raise TypeError("Input must be a sring")
    return text [::-1] 

def capitalize_string(text): 
    if not isinstance(text, str): 
        raise TypeError ("Input must be a string") 
    
    return text.capitalize() 

def to_uppercase(text): 
    if not isinstance(text, str): 
        raise TypeError("Inpput must be a string") 
    return text.upper() 

def to_lowercase(text): 

    if not isinstance(text, str): 
        raise TypeError ("Input must be a string") 
    return text.lower() 


print(reverse_string("python")) 

print(capitalize_string("cm")) 

print (to_uppercase("chaitmvn"))

print(to_lowercase("OMNERCAT"))
```

```python
# The code for the following program is as :   

def count_words_in_file(filename): 
    try: 
        with open(filename, 'r') as file : 
            content = file.read() 
            words = content.split() 
            return len(words) 
        
    except FileNotFoundError: 
        print("File not found error")
        return 0 
    
filename = "data.txt" 
word_count = count_words_in_file(filename) 

print(f"Total number of wwords in '{filename} : {word_count}")
```

Develop a Python program that reads a text file named "contacts.txt" and extracts email addresses from
each line of the file0

```python
# the code for the following program is as : 

import re 
def extract_emails(filename): 
    email_pattern = r'[a- za- z0-9-_%+-]+@[a-zA-z0-9.-]+\.[a-za-z]{2,}'
    emails = {} 

    try : 
        with open(filename, 'r') as file: 
            for line in file : 
                found_emails = re.findall(email_pattern, line) 
                emails.extend(found_emails) 

            return emails 
        
    except FileNotFoundError : 
        print("File does not exist")
        return []
    
emails = extract_emails("contacts.txt") 

if emails : 
    print("Email address found")
    for email in emails : 
        print(email) 

else :
    print("No email address found")
```

Q11.    
Write a Python program to open a text file named "grades.txt" and find the average grade of students
listed in the file0

```python
# the code for the above example is as follows :  

def calculate_average_grade(): 
    try : 
        with open("grades.txt", "r") as file : 
            grades = [] 

            for line in file : 
                line = line.strip() 
                if line : 
                    grades.append(float(line)) 

            if grades : 
                average = sum(grades) / len(grades) 
                print("Average Grade: ", average) 

            else: 
                print("No grades found in the file") 

    except FileNotFoundError : 
        print("The file 'grades.txt does not exist") 

    except ValueError : 
        print("Please ensure that all the grades in the file are numeric") 

calculate_average_grade()
```

Q12.   
Create a Python script that reads a text file named "expenses.txt" and calculates the total amount spent
on various expenses listed in the file.

```python
# The code for the given example is : 
total_expense = 0.0 

try : 
    with open("expenses.txt", "r") as file : 
        for line in file : 
            parts = line.strip().split() 
            if len(parts) >= 2 : 
                amount = float(parts[-1])
                total_expense += amount 

    print(f'Total amount spent: {total_expense}') 

except FileNotFoundError : 
    print("Expenses not found") 

except ValueError : 
    print("Invalid error not found")
```
