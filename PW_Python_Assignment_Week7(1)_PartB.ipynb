Q1. 
Implement a temperature converter program in Python that converts Celsius to Fahrenheit and vice versa. However, include exception handling to handle cases where the user enters invalid input (e.g., non-numeric values or out-of-range temperatures). 

```python
# The code for the following program is as follows : 

def celsius_to_fahrenheit(celsius): 
    fahrenheit = (celsius * 9/5) + 32 
    return fahrenheit 

def fahrenheit_to_celsius(fahrenheit): 
    celsius = (fahrenheit - 32) * 5/9
    return celsius 

def main(): 
    while True : 
        try : 
            choice = input("Enter 'C' to convert Celsius to Fahrenheit, 'F' to convert Fahrenheit to Celsius or Q to quit: " ).upper() 

            if choice == 'C' : 
                break 

            temperature = float(input("Enter the temperature: "))

            if choice == "C" : 
                converted_temp = celsius_to_fahrenheit(temperature)
                print(f'{temperature} degree celsius is equal to {converted_temp} degree Fahrenheit')

            elif choice == 'F' : 
                converted_temp = fahrenheit_to_celsius(temperature)
                print(f'{temperature} degree fahrenheit is equal to {converted_temp} degree celsius')

            else: 
                print("Invalid choice. Please try again")

        except ValueError : 
            print("Invalid Input. Please enter a valid number")

if __name__ == "__main__": 
    main()
```

Design a student grading system using Python where users can input student names and their corresponding grades. Ensure to handle exceptions for invalid input grades (e.g., grades outside the valid range of 0 to 100).

```python
# The code for the above example is as : 
class Student : 
    def __init__(self,name,grade): 
        self.name = name
        self.grade = self.validate_grade(grade) 

    def validate_grade(self, grade): 
        try: 
            grade = float(grade)
            if 0 <= grade <= 100 : 
                return grade
            else : 
                raise ValueError("Grade must be between 0 and 100")
            
        except ValueError as  e : 
            print(f"Invalid grade for {self.name} : {e}")
            return None
        

class GradingSystem : 
        def __init__(self): 
            self.students = []

        def add_student(self, name, grade): 
            student = Student(name,grade)
            if student.grade is not None : 
                self.students.append(student)

        def display_grades(self): 
            if not self.students : 
                print("No students added yet")

            else: 
                print('Students Grades')
                for student in self.students: 
                    print(f"{student.name} : {student.grade}")


# Example Usage 
grading_system = GradingSystem()

while True : 
    name = input("Enter the student name")
    if name.lower() == 'q' : 
        break

    grade = input("Enter the Student Grade: ")
    grading_system.add_student(name, grade)

grading_system.display_grades()


            
        
```

Build a basic banking system simulator in Python where users can deposit and withdraw money from their accounts. Implement exception handling to prevent overdrafts (withdrawals exceeding the account balance) and handle invalid input.

```python
# The code for the above example is as : 
class BankAccount : 
    def __init__(self,  name, balance = 0.0): 
        self.name =  name
        self.balance =  balance 

    def deposit(self, amount): 
        try : 
            amount = float(amount)
            if amount <= 0: 
                raise ValueError ('Depsoit Amount must be greater than the zero')
            self.balance += amount
            print(f"Deposited ${amount:.2f} into {self.name}'s account. New Balance ${self.balance}")

        except ValueError as e : 
            print(f"Invalid deposit Amount: {e}")

    def withdraw(self, amount): 
        try: 
            amount = float(amount)
            if amount <= 0: 
                raise ValueError("Withdrawl amount mut be greater than the zero")
            
            if amount > self.balance : 
                raise ValueError("Insufficient balance Cannot withdraw more thanavailable balance") 
            self.balance -= amount
            print(f"Withdraw ${amount:.2f} from {self.name}'s account New balance : ${self.balance:.2f}")
        except ValueError as e : 
            print(f"Invalid withdrawl amount : {e}")

    def display_details(self): 
        print(f"{self.name}'s account balance: ${self.balance}")

# Example Usage 
account = BankAccount("Chaitanya mangla", 1000.0)

while True : 
    choice = input("Enter 'd' to deposit, 'w'to withdraw, 'b' to check balanceor 'q' to quit: ")

    if choice.lower() == 'd' : 
        amount = input("Enter the deposit amoount")
        account.deposit(amount)

    elif choice.lower() == 'w' : 
        amount = input("Enter the withdrawl amount: ")
        account.withdraw(amount)

    elif choice.lower() == 'b' : 
        account.display_details()

    elif choice.lower() == 'q' : 
        break 

    else : 
        print("Invalid choice please try again")
```

```python
# The code for the above example is as : 

import math 

def calculate(operation, num1, num2 = None): 
    try : 
        if operation == '+' :
            result = num1 + num2 
        elif operation == '-' : 
            result = num1 - num2 
        elif operation == '*' : 
            result = num1 * num2 
        elif operation == '/' : 
            if num2 == 0:
                raise ValueError ("Cannot divide by Zero")
            result = num1 / num2 
        
        elif operation == 'sqrt' : 
            if num1 < 0 : 
                raise ValueError ("Cannot take square root of a negative number")
            result = math.sqrt(num1)

        elif operation == 'log' : 
            if num1 <= 0 : 
                raise ValueError ("Logarithm is undefined for unpositive numbers")
            result = math.log(num1)

        elif operation == 'exp' : 
            result = math.exp(num1)

        else: 
            raise ValueError ("Invalid Operation")
        return result
    
    except ValueError as e : 
        return str(e)
    
def main(): 
    print("Scientific Calculator")
    print("Available Operations: +, -, *, /, sqrt, log, exp")
    print("For QUIT : quit")

    while True : 
        operation = input("Enter Operation").lower()
        if operation == 'quit' : 
            break 

        if operation in ['+', '-', '*',  '/']: 
            num1 = float(input("Enter the first number: "))
            num2 = float(input("Enter the second number: "))
            result = calculate(operation, num1, num2)
        else : 
            num1 == float(input("Enter number: "))
            result = calculate(operation, num1)

        print(f"Result: {result}")
        print()

if __name__ == "__main__" : 
    main()
```

Implement a Polynomial class in Python that supports arithmetic operations on polynomials using operator overloading. The class should allow users to define polynomials of any degree and perform addition, subtraction, and multiplication operations on them.

```python
class Polynomial:
    def __init__(self, coefficients):
        self.coefficients = coefficients

    def __repr__(self):
        terms = []
        for i, coeff in enumerate(self.coefficients):
            if coeff != 0:
                if i == 0:
                    term = str(coeff)
                elif i == 1:
                    term = f"{coeff}x"
                else:
                    term = f"{coeff}x^{i}"
                terms.append(term)
        return " + ".join(terms) or "0"

    def __add__(self, other):
        if isinstance(other, Polynomial):
            max_degree = max(len(self.coefficients), len(other.coefficients))
            result_coefficients = [0] * max_degree

            for i, coeff in enumerate(self.coefficients):
                result_coefficients[i] += coeff

            for i, coeff in enumerate(other.coefficients):
                result_coefficients[i] += coeff

            return Polynomial(result_coefficients)
        else:
            return NotImplemented

    def __sub__(self, other):
        if isinstance(other, Polynomial):
            return self + (-1 * other)
        else:
            return NotImplemented

    def __mul__(self, other):
        if isinstance(other, Polynomial):
            result_coefficients = [0] * (len(self.coefficients) + len(other.coefficients) - 1)

            for i, coeff1 in enumerate(self.coefficients):
                for j, coeff2 in enumerate(other.coefficients):
                    result_coefficients[i + j] += coeff1 * coeff2

            return Polynomial(result_coefficients)
        elif isinstance(other, (int, float)):
            result_coefficients = [coeff * other for coeff in self.coefficients]
            return Polynomial(result_coefficients)
        else:
            return NotImplemented

    def __neg__(self):
        return Polynomial([-coeff for coeff in self.coefficients])

    def __rmul__(self, other):
        return self * other

# Example usage
p1 = Polynomial([2, 3, -1])  # 2 + 3x - x^2
p2 = Polynomial([1, -2, 1])  # 1 - 2x + x^2

print("p1:", p1)
print("p2:", p2)

p3 = p1 + p2
print("p1 + p2:", p3)

p4 = p1 - p2
print("p1 - p2:", p4)

p5 = p1 * p2
print("p1 * p2:", p5)

p6 = p1 * 2
print("p1 * 2:", p6)

p7 = 2 * p1
print("2 * p1:", p7)
```
